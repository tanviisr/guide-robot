```python
import RPi.GPIO as GPIO
import time

# GPIO pin setup
IN1 = 17  # Motor A
IN2 = 18
IN3 = 22  # Motor B
IN4 = 23

# Set GPIO mode
GPIO.setmode(GPIO.BCM)
GPIO.setwarnings(False)

# Setup motor pins
GPIO.setup(IN1, GPIO.OUT)
GPIO.setup(IN2, GPIO.OUT)
GPIO.setup(IN3, GPIO.OUT)
GPIO.setup(IN4, GPIO.OUT)

# Movement functions
def stop():
    GPIO.output(IN1, 0)
    GPIO.output(IN2, 0)
    GPIO.output(IN3, 0)
    GPIO.output(IN4, 0)
def forward():
    GPIO.output(IN1, 1)
    GPIO.output(IN2, 0)
    GPIO.output(IN3, 1)
    GPIO.output(IN4, 0)

def backward():
    GPIO.output(IN1, 0)
    GPIO.output(IN2, 1)
    GPIO.output(IN3, 0)
    GPIO.output(IN4, 1)

def left():
    GPIO.output(IN1, 0)
    GPIO.output(IN2, 1)
    GPIO.output(IN3, 1)
    GPIO.output(IN4, 0)
def right():
    GPIO.output(IN1, 1)
    GPIO.output(IN2, 0)
    GPIO.output(IN3, 0)
    GPIO.output(IN4, 1)

# Main loop
try:
    print("Moving forward")
    forward()
    time.sleep(3)

    print("Turning left")
    left()
    time.sleep(1)

    print("Moving backward")
    backward()
    time.sleep(3)

    print("Stopping")
    stop()

except KeyboardInterrupt:
    print("Program stopped by user")
    stop()
    GPIO.cleanup()
```
