```python
import pyttsx3
import speech_recognition as sr
import json

DATA_PATH = "Downloads/file1.txt"  # Your robot data file path

engine = pyttsx3.init()
engine.setProperty('rate', 150)

def speak(text):
    engine.say(text)
    engine.runAndWait()

def load_robot_data():
    with open(DATA_PATH, "r") as file:
        return json.load(file)
def introduce_robot():
    intro = (
        "Hello! I am your Department Guide Robot. "
        "I can tell you about Kinova Lite, AMR100, Transmission Line Robot, and Nao Robot. "
        "Please say the name of the robot you'd like to know more about."
    )
    print("??", intro)
    speak(intro)

def recognize_speech():
    recognizer = sr.Recognizer()
    
    # ?? Change this to the correct device index of your USB mic
    mic_index = 1  # ?? Change this after testing with list_microphone_names()
    
    with sr.Microphone(device_index=mic_index) as source:
        print("?? Listening...")
        recognizer.adjust_for_ambient_noise(source, duration=1)
        speak("Please say the name of the robot.")
        audio = recognizer.listen(source, timeout=30)

    try:
        query = recognizer.recognize_google(audio)
        print("? Recognized:", query)
        return query.lower()
    except sr.UnknownValueError:
        speak("Sorry, I couldn't understand that.")
        return None
    except sr.RequestError:
        speak("Speech recognition service is unavailable.")
        return None

def correct_robot_name(query):
    corrections = {
        "kinova": "kinova lite",
        "innova":"kinova",
        "amr": "amr100",
        "ar hundred": "amr100" ,
        "transmission": "transmission line robot",
        "nao": "nao robot",
        "now": "nao robot"
    }
    for wrong, correct in corrections.items():
        if wrong in query:
            return correct
    return query
def describe_robot(data, name):
    for robot in data:
        if robot["name"].lower() == name:
            description = f"{robot['name']} is manufactured by {robot['manufacturer']}. "
            if "weight_kg" in robot:
                description += f"It weighs {robot['weight_kg']} kg. "
            if "payload_kg" in robot:
                description += f"It can carry a payload of {robot['payload_kg']} kg. "
            if "degrees_of_freedom" in robot:
                description += f"It has {robot['degrees_of_freedom']} degrees of freedom. "
            if "power_source" in robot:
                description += f"It is powered by {robot['power_source']}. "
            if "functions" in robot:
                description += "Its key functions include " + ", ".join(robot["functions"]) + "."
            speak(description)
            print("?? Description:", description)
            return
    speak("Sorry, I could not find information about that robot.")

def main():
    robot_data = load_robot_data()
    introduce_robot()
    query = recognize_speech()
    if query:
        robot_name = correct_robot_name(query)
        describe_robot(robot_data, robot_name)
        else:
        speak("Please try again.")

if __name__ == "__main__":
    main()
```
